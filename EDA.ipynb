## Week 2: Data Cleaning and Preprocessing

#### 1. Clean messy column names

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
df.columns
```

#### 2. Remove duplicate rows

```python
print("Duplicates before:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
```

#### 3. Drop 'loanid' as it is not needed for Machine Learning

```python
if 'loanid' in df.columns:
    df.drop('loanid', axis=1, inplace=True)

print(df.columns)
```

#### 4. Detect and Handle Outliers (IQR Method)

```python
outlier_cols = ['income', 'loanamount']
for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    # only the rows within the lower and upper bounds
    df = df[(df[col] >= lower) & (df[col] <= upper)]

print("Outliers handled. New shape:", df.shape)
```

#### 5. Visualize Numeric & Categorical Columns

##### Visualize Numerical Data (Histogram & Box Plot)

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
df['age'].hist(bins=20)
plt.title('Histogram of Age')

plt.subplot(1, 2, 2)
sns.boxplot(x=df['loanamount'])
plt.title('Box Plot of Loan Amount')
plt.show()
```

##### Visualize Categorical Data (Count Plot & Pie Chart)

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.countplot(x='education', data=df)
plt.title('Count Plot of Education')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
df['employmenttype'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Pie Chart of Employment Type')
plt.ylabel('')
plt.show()
```

#### 6. Encode Categorical Columns

```python
categorical_cols = df.select_dtypes(include=['object']).columns
df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
bool_cols = df.select_dtypes(include=['bool']).columns
df[bool_cols] = df[bool_cols].astype(int)

print("Categorical columns encoded successfully.")
```

#### 7. Scale Numeric Columns

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
cols_to_scale = ['age', 'income', 'loanamount', 'creditscore', 'monthsemployed', 
                 'numcreditlines', 'interestrate', 'loanterm', 'dtiratio']
df[cols_to_scale] = scaler.fit_transform(df[cols_to_scale])

print("Numeric columns scaled.")
df.head()
```
